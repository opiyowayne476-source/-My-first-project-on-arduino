```python
# ==========================
# DATA CLEANING PROJECT
# ==========================

import pandas as pd
import matplotlib.pyplot as plt

# 1. LOAD DATA
data = pd.read_csv("student.csv")

# 2. VIEW DATA
print("FIRST 5 RECORDS")
print(data.head())

# 3. CHECK DATA TYPES
print("\nDATA TYPES")
print(data.dtypes)

# 4. CHECK MISSING VALUES
print("\nMISSING VALUES")
print(data.isnull().sum())

# 5. REMOVE DUPLICATES
print("\nNUMBER OF DUPLICATES:", data.duplicated().sum())

data = data.drop_duplicates()

# 6. CONVERT TO NUMERIC
data["Age"] = pd.to_numeric(data["Age"], errors="coerce")
data["Marks"] = pd.to_numeric(data["Marks"], errors="coerce")
data["Attendance"] = pd.to_numeric(data["Attendance"], errors="coerce")

# 7. FILL MISSING VALUES
data["Age"] = data["Age"].fillna(data["Age"].mean())
data["Marks"] = data["Marks"].fillna(data["Marks"].mean())
data["Attendance"] = data["Attendance"].fillna(data["Attendance"].mean())

# 8. DESCRIPTIVE STATISTICS
print("\nSUMMARY STATISTICS")
print(data.describe())

# 9. REMOVE OUTLIERS
data = data[data["Marks"] < 100]

# 10. SCATTER PLOT
plt.scatter(data["Age"], data["Marks"])
plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")
plt.show()

# 11. BAR GRAPH
data.groupby("Department")["Marks"].mean().plot(kind="bar")

plt.title("Average Marks per Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()

# 12. HISTOGRAM
data["Marks"].hist()

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

# 13. BOXPLOT
data.boxplot(column="Marks")

plt.title("Marks Boxplot")
plt.show()

# 14. SAVE CLEAN DATA
data.to_csv("clean_student.csv", index=False)

print("\nDATA CLEANING COMPLETED")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read all lines
with open("/content/student.CSV", "r") as f:
    lines = [line.strip() for line in f if line.strip()]

# Repair broken records
records = []
current = ""

for line in lines[2:]:     # Skip broken header lines
    if line.startswith("S"):
        if current:
            records.append(current)
        current = line
    else:
        current += " " + line

if current:
    records.append(current)

# Create dataframe
data = pd.DataFrame(
    [r.split(",", 6) for r in records],
    columns=[
        "Student_ID",
        "Name",
        "Age",
        "Gender",
        "Marks",
        "Attendance",
        "Department"
    ]
)

# Convert numeric columns
data["Age"] = pd.to_numeric(data["Age"], errors="coerce")
data["Marks"] = pd.to_numeric(data["Marks"], errors="coerce")
data["Attendance"] = pd.to_numeric(data["Attendance"], errors="coerce")

# Missing values
print("Missing Values")
print(data.isnull().sum())

# Fill missing values
data["Age"] = data["Age"].fillna(data["Age"].mean())
data["Marks"] = data["Marks"].fillna(data["Marks"].mean())
data["Attendance"] = data["Attendance"].fillna(data["Attendance"].mean())

# Duplicates
print("\nDuplicates:", data.duplicated().sum())

# Remove duplicates
data = data.drop_duplicates()

# Remove outlier (450 marks)
data = data[data["Marks"] <= 100]

# Summary statistics
print("\nSummary")
print(data.describe())

# First records
print("\nClean Data")
print(data.head())

# Scatter Plot
plt.scatter(data["Age"], data["Marks"])
plt.xlabel("Age")
plt.ylabel("Marks")
plt.title("Age vs Marks")
plt.show()

# Bar Graph
data.groupby("Department")["Marks"].mean().plot(kind="bar")
plt.title("Average Marks Per Department")
plt.ylabel("Marks")
plt.show()

# Histogram
plt.hist(data["Marks"], bins=10)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

# Boxplot
plt.boxplot(data["Marks"])
plt.title("Marks Boxplot")
plt.show()

# Save cleaned file
data.to_csv("/content/clean_student.csv", index=False)

print("\nClean file saved as clean_student.csv")
```
